```python
import pandas as pd
import numpy as np
```

```python
beml_df = pd.read_csv('/content/Data-Science/Statistics/BEML.csv')
beml_df.head()
```

```python
glaxo_df = pd.read_csv('/content/Data-Science/Statistics/GLAXO.csv')
glaxo_df.head()
```

```python
beml_df = beml_df[['Date','Close']]
glaxo_df = glaxo_df[['Date','Close']]
```

```python
beml_df
```

```python
beml_df = beml_df.set_index(pd.DatetimeIndex(beml_df['Date']))
glaxo_df = glaxo_df.set_index(pd.DatetimeIndex(glaxo_df['Date']))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.plot(glaxo_df.Close)
plt.xlabel("Time")
plt.ylabel("Closed Price")
```

```python
plt.plot(beml_df.Close)
plt.xlabel("Time")
plt.ylabel("Closed Price")
```

```python
glaxo_df["gain"] = glaxo_df.Close.pct_change(periods = 1)
beml_df["gain"] = beml_df.Close.pct_change(periods = 1)
```

```python
beml_df
```

```python
glaxo_df
```

```python
beml_df = beml_df.dropna()
glaxo_df = glaxo_df.dropna()
```

```python
plt.figure(figsize = (8,6))
plt.plot(glaxo_df.index,glaxo_df.gain)
plt.xlabel("Time")
plt.ylabel("Gain")
```

```python
sns.distplot(glaxo_df.gain, label = "Glaxo")
plt.xlabel("Gain")
plt.ylabel("Density")
plt.legend()
```

```python
sns.distplot(beml_df.gain, label = "BEML")
plt.xlabel("Gain")
plt.ylabel("Density")
plt.legend()
```

```python
print("Mean:", round(glaxo_df.gain.mean(),4))
print("Standard Deviation:", round(glaxo_df.gain.std(),4))
```

```python
print("Mean:",round(beml_df.gain.mean(),4))
print("Standard Deviation:",round(beml_df.gain.std(),4))
```

```python
# probablity of making 2% loss or higher in glaxo
stats.norm.cdf(-0.02, loc=glaxo_df.gain.mean(), scale=glaxo_df.gain.std())
```

```python
# probablity of making 2% gain or higher in glaxo
1-stats.norm.cdf(0.02, loc=glaxo_df.gain.mean(), scale=glaxo_df.gain.std())
```

```python
# probablity of making 2% loss or higher in beml
stats.norm.cdf(-0.02,loc=beml_df.gain.mean(),scale=beml_df.gain.std())
```

```python
# probablity of making 2% gain or higher in beml
1-stats.norm.cdf(0.02,loc=beml_df.gain.mean(),scale=beml_df.gain.std())
```
