# One tail test on Icecream Salmonella levels

```python
data=pd.Series([0.593, 0.142, 0.329, 0.691, 0.231, 0.793, 0.519, 0.392, 0.418])
```

```python
stats.ttest_1samp(data,0.3)
```

```python
p_value = stats.ttest_1samp(data,0.3)[1]
p_value
```

```python
p = p_value/2
p
```

```python
# p < alpha         (reject Null hypothesis)
```

# 2 proportion test

```python
#Data:

n1 = 247
p1 = .37

n2 = 308
p2 = .39
```

```python
population1 = np.random.binomial(1, p1, n1)
population2 = np.random.binomial(1, p2, n2)
```

```python
population1
```

```python
population1
```

```python
population1.mean()
```

```python
population2.mean()
```

```python
import statsmodels.api as sm
sm.stats.ttest_ind(population1,population2)
```

```python
stats.ttest_ind(population1,population2)
```

```python
0.950 > 0.05      # Reject Alternative hypothesis
```

```python
# So both proportions are equal (Accept null hypothesis)
```
