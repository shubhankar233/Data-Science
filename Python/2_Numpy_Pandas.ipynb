6.	Given a 'marks' variable, print the grade:
- 90 and above: 'A'
- 75–89: 'B'
- 50–74: 'C'
- Below 50: 'Fail'


```python
marks = int(input("Enter the marks: "))
if marks >= 90:
    print("A")
elif marks >= 75 and marks <= 89:
    print("B")
elif marks >= 50 and marks <= 74:
    print("C")
else:
    print("Fail")
```

7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
x = int(input("Enter the number: "))
if x > 0:
    if x%2 == 0:
        print(f"{x} is positive and even")
    else:
        print(f"{x} is positive")
elif x == 0:
    print("Zero")
else:
    print(f"{x} is negative")
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
data = {
    'Name': ['Rahul', 'Amit', 'Sneha', 'Priya', 'Karan',
             'Neha', 'Arjun', 'Riya', 'Rahul', 'Sneha'],
    'Age': [25, 30, 28, 22, 35,
            27, 29, 24, 25, 28],
    'City': ['Mumbai', 'Delhi', 'Pune', 'Mumbai', 'Delhi',
             'Chennai', 'Pune', 'Mumbai', 'Mumbai', 'Pune'],
    'Salary': [50000, 60000, None, 45000, 70000,
               52000, None, 48000, 50000, None]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.info()
```

```python
df.describe()
```

```python
df[['Name','City']]
```

```python
df.drop('City', axis=1, inplace=True)          # inplace = True means it will change the original data aswell
df
```

```python
df.Salary.fillna(df.Salary.mean(), inplace=True)
df
```

```python
df.drop_duplicates(inplace=True)
df
```
